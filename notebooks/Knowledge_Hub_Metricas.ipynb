# Evaluación de Softtek Knowledge Hub

Esta evaluación reúne los resultados obtenidos en las pruebas del prototipo. Cada función se mide con una tarea distinta: asignar una categoría, encontrar el documento esperado, recomendar documentos relacionados o responder con respaldo en una fuente.

Los conteos provienen de las ejecuciones documentadas en los notebooks anteriores. Esta celda calcula los porcentajes a partir de esos resultados.

```python
import pandas as pd
from IPython.display import display

# Cada fila conserva el número de casos que se evaluaron en esa prueba.
filas = [
    ("Clasificación", "Original · validación cruzada", "Documentos bien clasificados", 54, 60),
    ("Clasificación", "Combinado · validación cruzada", "Documentos bien clasificados", 57, 60),
    ("Clasificación", "Combinado · umbral del 40%", "Aciertos entre los aceptados", 41, 42),
    ("Clasificación", "Combinado · umbral del 40%", "Documentos aceptados automáticamente", 42, 60),
    ("Búsqueda en español", "TF-IDF", "Documento esperado en primer lugar", 1, 10),
    ("Búsqueda en español", "E5 multilingüe", "Documento esperado en primer lugar", 9, 10),
    ("Recomendación", "IR01 · similitud textual", "Vínculos conocidos recuperados en top 5", 1, 2),
    ("Recomendación", "IR01 · ontología + similitud", "Vínculos conocidos recuperados en top 5", 2, 2),
    ("Recomendación", "IR07 · evaluación manual", "Sugerencias relevantes en top 5", 1, 5),
    ("Ontología", "Corpus completo", "Documentos con una ruta de recomendación", 6, 60),
]

resumen = pd.DataFrame(
    filas, columns=["Componente", "Prueba", "Qué se midió", "Conteo", "Total"]
)
resumen["Resultado"] = (100 * resumen["Conteo"] / resumen["Total"]).map("{:.1f}%".format)
display(resumen)
```

Los dos casos prueban comportamientos distintos con una misma fuente: responder cuando el dato aparece y reconocer cuándo falta. Queda pendiente evaluar la calidad de las respuestas con documentos de otras categorías.
